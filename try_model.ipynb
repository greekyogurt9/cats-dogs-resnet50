```python
import tensorflow
from tensorflow import keras
```

```python
model = keras.models.load_model(r"C:\Users\ankit\Downloads\cats_dogs_resnet50.keras")
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

image_path = r"C:\Users\ankit\Desktop\images.jpg"

image = tf.keras.utils.load_img(
    image_path,
    target_size=(224, 224)
)

image_array = tf.keras.utils.img_to_array(image)

image_batch = np.expand_dims(image_array, axis=0)

prediction = model.predict(image_batch, verbose=0)[0][0]

if prediction >= 0.5:
    label = "dog"
else:
    label = "cat"

print(f"Prediction: {label}")
print(f"Dog probability: {prediction:.2%}")

plt.imshow(image)
plt.axis("off")
plt.show()
```
